```python
# Część 1: Wczytaj dane

from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("Lab2-Transactions")
    .getOrCreate()
)
spark.sparkContext.setLogLevel("WARN")
print(f"Spark {spark.version} — gotowy")
```

```python
df = spark.read.json("data/transactions_10k.jsonl")

print(f"Liczba rekordów: {df.count()}")
df.printSchema()
```

```python
df.show(10, truncate=False)
```

```python
from pyspark.sql.functions import to_timestamp, col

df = df.withColumn("timestamp", to_timestamp(col("timestamp"), "yyyy-MM-dd HH:mm:ss"))

df.printSchema()  # timestamp powinien być teraz 'timestamp (nullable = true)'
```

```python
# Część 2: Podstawowe agregacje
# Zadanie 2.1 — Liczba transakcji i suma przychodów per sklep


from pyspark.sql.functions import count, sum as _sum, avg, round as _round

store_summary = (
    df.groupBy("store")
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
        _round(avg("amount"), 2).alias("srednia_PLN"),
    )
    .orderBy("store")
)
store_summary.show()
```

```python
# Zadanie 2.2 - Statystyki per kategoria

from pyspark.sql.functions import min as _min, max as _max

# TWÓJ KOD
df.groupBy("category").agg(_round(_sum("amount"), 2).alias("suma_PLN"),_round(_min("amount"), 2).alias("min_PLN"),_round(_max("amount"), 2).alias("max_PLN")).orderBy("category").show()
```

```python
# Część 3: Okna tumbling
# Zadanie 3.1 — Liczba transakcji per godzina (tumbling 1h)


from pyspark.sql.functions import window

hourly = (
    df.groupBy(window("timestamp", "1 hour"))    # okno 1-godzinne
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .orderBy("window")
)
hourly.show(truncate=False)
```

```python
(
    hourly
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .show(truncate=False)
)
```

```python
# Zadanie 3.2 Okna 30-minutowe per sklep

(
    df.groupBy(window("timestamp", "30 minutes"), "store")
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN")
    )
    .orderBy("window", "store")
    .show(truncate=False)
)
```

```python
# Zadanie 3.3 — W której godzinie sklep “Kraków” miał najwyższy przychód?

from pyspark.sql.functions import desc

(
    df.filter(col("store") == "Kraków")
    .groupBy(window("timestamp", "1 hour"))
    .agg(_round(_sum("amount"), 2).alias("suma_PLN"))
    .orderBy(desc("suma_PLN"))
    .show(1, truncate=False)
)
```

```python
# Część 4: Okna sliding
# Zadanie 4.1 — Okno 1h, krok 30 minut

sliding = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))  # szerokość 1h, krok 30min
    .agg(
        count("tx_id").alias("liczba_tx"),
        _round(_sum("amount"), 2).alias("suma_PLN"),
    )
    .select(
        col("window.start").alias("od"),
        col("window.end").alias("do"),
        "liczba_tx",
        "suma_PLN",
    )
    .orderBy("od")
)
sliding.show(truncate=False)
```

```python
# Zadanie 4.2 — Porównaj tumbling vs sliding

tumbling_rows = (
    df.groupBy(window("timestamp", "1 hour"))
    .agg(count("tx_id"))
    .count()
)
sliding_rows = (
    df.groupBy(window("timestamp", "1 hour", "30 minutes"))
    .agg(count("tx_id"))
    .count()
)
print(f"Tumbling (1h):          {tumbling_rows} okien")
print(f"Sliding  (1h / 30min):  {sliding_rows} okien")

# Dlaczego sliding ma więcej wierszy?
# Wynika to z nakładania się na siebie przedziałów czasowych.
# Przy oknie 1h z przesunięciem co 30 min, okna nakładają się na siebie, co generuje więcej okien wynikowych w tym samym czasie.
```

```python
# Część 5: Pytania kontrolne

# 1. Ile transakcji jest w oknie 09:00-10:00?

# liczba_tx - 4661
# +-------------------+-------------------+---------+----------+
# |od                 |do                 |liczba_tx|suma_PLN  |
# +-------------------+-------------------+---------+----------+
# |2026-04-12 09:00:00|2026-04-12 10:00:00|4661     |1896230.21|
# +-------------------+-------------------+---------+----------+


# 2. Jaka jest różnica między groupBy("store") a groupBy(window(...), "store")?
# groupBy("store") grupuje dane dla danego sklepu ze wszystkich rekordów naraz. 
# groupBy(window(...), "store") najpierw grupuje według przedziałów czasowych, a następnie wewnątrz każdego przedziału oddzielnie tworzy grupy dla poszczególnych sklepów.


# 3. W oknie sliding 1h/30min - ile okien zawiera transakcje z godziny 09:30?
# Transakcje z godziny 09:30 będą zawierały 2 okna - od 09:00 do 10:00 oraz od 09:30 do 10:30. Okna są przedziałami zamkniętymi lewostronnie.
```

```python
# Praca domowa
from pyspark.sql.functions import hour, minute

# 1. Gdańsk - najniższa średnia kwota
(
    df.filter(col("store") == "Gdańsk")
    .groupBy(window("timestamp", "1 hour"))
    .agg(_round(avg("amount"), 2).alias("srednia_kwota"))
    .orderBy("srednia_kwota")
    .show(1, truncate=False)
)

# 2. Liczba transakcji per kategoria w oknie 09:00-09:30
(
    df.filter((hour(col("timestamp")) == 9) & (minute(col("timestamp")) < 30))
    .groupBy("category")
    .agg(count("tx_id").alias("liczba_tx"))
    .show()
)

# 3. Szczyt transakcji w oknie 15-minutowym
(
    df.groupBy(window("timestamp", "15 minutes"))
    .agg(count("tx_id").alias("liczba_tx"))
    .orderBy(desc("liczba_tx"))
    .show(1, truncate=False)
)
```
